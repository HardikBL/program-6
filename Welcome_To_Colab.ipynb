```
import pandas as pd
import plotly.express as px

# Skip the first row containing 'file_name=website_traffic.csv'
df = pd.read_csv("website_traffic.csv", skiprows=1)

df["Date"] = pd.to_datetime(df["Date"])

fig = px.line(
    df,
    x="Date",
    y="Visitors",
    title="Website Traffic Analysis",
    hover_data=["Visitors"]
)

fig.update_traces(
    mode="lines+markers",
    line=dict(color="blue", width=4),
    marker=dict(color="red", size=10)
)

fig.update_xaxes(tickangle=45)

fig.update_layout(
    xaxis_title="Date",
    yaxis_title="Number of Visitors",
    template="plotly_white"
)
fig.write_html("Website.png")
fig.show()
```

```
import pandas as pd
import plotly.express as px

df = pd.read_csv("website_traffic.csv")
print(df.head())

df["Date"] = pd.to_datetime(df["Date"])

df = df.dropna()

fig = px.line(df, x="Date", y="Number of Website Visitors",
              title="Website Traffic Analysis Over Time",
              labels={"Date": "Date", "Number of Website Visitors": "Visitors"},
              hover_data=["Page Views", "Average Session Duration (minutes)",
                          "Bounce Rate (%)"])

fig.update_traces(mode="lines+markers",
                  line=dict(color="blue", width=3),
                  marker=dict(size=8, color="red"))

fig.update_layout(xaxis_tickangle=45,
                  template="plotly_white",
                  hovermode="x unified")

fig.write_html("website_traffic.html")
fig.show()
```
